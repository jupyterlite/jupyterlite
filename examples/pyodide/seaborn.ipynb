# Seaborn

> adapted from https://seaborn.pydata.org/examples/grouped_violinplots.html

```python
import seaborn as sns
import pyodide_http
```

Fetch the example data as a `pandas.DataFrame` from [seaborn-data](https://github.com/mwaskom/seaborn-data), patching the underlying HTTP request.

```python
pyodide_http.patch_all()
tips = sns.load_dataset("tips")
tips
```

Plot the dataset as violin plot in the SVG format.

```python
%config InlineBackend.figure_format = 'svg'
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="smoker",
    split=True,
    inner="quart",
    fill=False,
    palette={"Yes": "g", "No": ".35"},
);
```
